6) Add a new column 'Bonus' that is 10% of Salary and a column 'Total' = Salary + Bonus. Then rename the column 'City' to 'Location' and drop the 'Bonus' column.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Usman", "Ayesha", "Bilal"],
    "Age": [25, 32, 28, 40, 35, 22],
    "City": ["Karachi", "Delhi", "Lahore", "Delhi", "Karachi", "Delhi"],
    "Salary": [30000, 45000, 35000, 60000, 50000, 28000]
}

df = pd.DataFrame(data)
df["Bonus"] = df["Salary"] * 0.10

df["Total"] = df["Salary"] + df["Bonus"]

df = df.rename(columns={"City": "Location"})

df = df.drop(columns=["Bonus"])

print(df)
```

7) Create a DataFrame having some missing values (NaN). Find the count of missing values in each column, fill numeric NaNs with the column mean, and drop rows that still have missing values.

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Age": [22, np.nan, 30, 28],
    "Salary": [25000, 35000, np.nan, 40000],
    "City": ["Karachi", "Lahore", np.nan, "Delhi"]
}

df_missing = pd.DataFrame(data)

print("Original DataFrame:")
print(df_missing)

print("Missing Values in Each Column:")
print(df_missing.isnull().sum())

df_missing["Age"] = df_missing["Age"].fillna(df_missing["Age"].mean())

df_missing["Salary"] = df_missing["Salary"].fillna(df_missing["Salary"].mean())

print("After Filling Numeric Missing Values:")
print(df_missing)

df_missing = df_missing.dropna()

print("After Dropping Remaining Missing Values:")
print(df_missing)
```

8) Sort the DataFrame by Salary in descending order, and then by Age in ascending order. Also print the top 3 highest-paid employees using nlargest().

```python
import pandas as pd

data = {
    "Name": ["Ali", "Ahmed", "Sara", "Usman", "Ayesha", "Bilal"],
    "Age": [25, 32, 28, 40, 35, 22],
    "City": ["Karachi", "Delhi", "Lahore", "Delhi", "Karachi", "Delhi"],
    "Salary": [30000, 45000, 35000, 60000, 50000, 28000]
}

df = pd.DataFrame(data)
print("Sorted by Salary Descending and Age Ascending:")

sorted_df = df.sort_values(
    by=["Salary", "Age"],
    ascending=[False, True]
)

print(sorted_df)

print("Top 3 Highest-Paid Employees:")
print(df.nlargest(3, "Salary"))
```

9) Using groupby, find the average salary per City, the number of employees per City, and the maximum salary per City. Show the result using agg() with multiple functions.

```python
result = df.groupby("City")["Salary"].agg(
    ["mean", "count", "max"]
)

print(result)
```

10) Create two DataFrames (employees and departments) with a common column 'DeptID'. Merge them using inner and left joins, then concatenate two DataFrames of the same structure using pd.concat() and remove duplicate rows.

```python
import pandas as pd

employees = pd.DataFrame({
    "Name": ["Ali", "Sara", "Ahmed", "Bilal"],
    "DeptID": [1, 2, 1, 3]
})

departments = pd.DataFrame({
    "DeptID": [1, 2, 4],
    "Department": ["IT", "HR", "Finance"]
})

print("Inner Join:")
print(pd.merge(employees, departments, on="DeptID", how="inner"))

print("Left Join:")
print(pd.merge(employees, departments, on="DeptID", how="left"))

employees2 = pd.DataFrame({
    "Name": ["Ali", "Sara", "Usman"],
    "DeptID": [1, 2, 3]
})

combined = pd.concat([employees, employees2], ignore_index=True)

combined = combined.drop_duplicates()

print("After Concatenation and Removing Duplicates:")
print(combined)
```
