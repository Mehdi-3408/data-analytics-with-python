Q4 — Count Frequency of Numbers

numbers = [1,2,2,3,3,3,4,4,4,4]

Create frequency dictionary.

```python
# creating frequency dictionary
numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for i in numbers:
    if i in frequency:
        frequency[i] += 1
    else:
        frequency[i] = 1
print(frequency)
```

Q5 — Word Counter

Take sentence input and count word frequency.

```python
word = "hello! how are you?".split()
freq = {}

for i in word:
    if i in freq:
        freq[i] += 1
    else:
        freq[i] = 1
print(freq)
```

Q6 — Student Marks & Average

marks = {"Math":80, "English":70, "Physics":90}

Calculate total and average.

```python
marks = {"Math":80, "English":70, "Physics":90}

for mark in marks.values():
    total = sum(marks.values())
    average = total / len(marks)
print(f"Total marks: {total} \n Average marks: {average}")
    
```

Q7 — Highest Scoring Student

students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}

Find highest scoring student.

```python
students = {
"Ali":85,
"Sara":92,
"Ahmed":78,
"Zara":95
}
highest_marks = 0
top_student = ""

for student, marks in students.items():
    if marks > highest_marks:
        highest_marks = marks
        top_student = student

print(f"Highest marks: {highest_marks} by {top_student}")
```

Q9 — Mini Project: Contact Book

Add 3 contacts, search by name, print phone or Contact not found.

```python
contact = {
    "Ali": {
        "phone": 1234567890
    },
    "Sara": {
        "phone": 9876543210
    }}
contact.update({"Ahmed": {"phone": 5555555555}})
contact.update({"Zara": {"phone": 1111111111}})
contact.update({"john": {"phone": 2222222222}})

name = input("Enter the name of the contact you want to search: ")
for name1, cntct_info in contact.items():
    if name == name1:
        print(f"Phone number of {name1}: {cntct_info['phone']}")
        break
    else:
        print(f"{name} not found in contacts.")
        break
```
