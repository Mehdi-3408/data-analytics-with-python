 3. Find and display the largest number of a list without using built-in function max(). Your program 
should ask the user to input values in list from keyboard.

```python
nums = list(map(int, input("Enter numbers: ").split()))
largest = nums[0] 

for num in nums:
    if num > largest:
        largest = num

print("The largest number is:", largest)
```

 4. Write a program that rotates the element of a list so that the element at the first index moves to the 
second index, the element in the second index moves to the third index, etc., and the element in the last 
index moves to the first index. 

```python
nums = list(map(int, input("Enter numbers: ").split()))

last = nums[-1]
first = nums[0]

nums[0] = last

for i in range(1, len(nums)):
    temp = nums[i]
    nums[i] = first
    first = temp

print("Rotated list:", nums)
```

5. Write a program that input a string and ask user to delete a given word from a string.  

```python
words = input("Enter words with space: ").split()
word = input("Enter word you want to delete: ")


for i in range(len(words)):
    if words[i] == word:
        words.remove(word)
        print("Word deleted successfully",word)

print("Updated list of words:", words)
```

8. Find the sum of each row of matrix of size m x n. For example for the following matrix output will be 
like this : 
 
Sum of row 1 = 32 
Sum of row 2 = 31 
Sum of row 3 = 63 

```python
mat = [[2, 11, 7, 12], [5, 2, 9, 15], [8, 3, 10, 42]]

for i in range(len(mat)):
    sum = 0
    for j in range(len(mat[i])):
        sum = sum + mat[i][j]
    print("Sum of row", i + 1, "is:", sum)
```

9. Write a program to add two matrices of size n x m. 

```python
a = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
b = [[9, 8, 7], [6, 5, 4], [3, 2, 1]]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j] + b[i][j], end=" ")
    print()
```

10. Write a program to multiply two matrices 

```python
a = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
b = [[9, 8, 7], [6, 5, 4], [3, 2, 1]]

for i in range(len(a)):
    for j in range(len(a[i])):
        print(a[i][j] * b[i][j], end=" ")
    print()
```
